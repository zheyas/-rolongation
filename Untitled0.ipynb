```python
import pandas as pd
import numpy as np
import re
from openpyxl import Workbook
from openpyxl.chart import BarChart, LineChart, Reference, PieChart
from openpyxl.styles import Font, PatternFill, Alignment, Border, Side
from openpyxl.utils import get_column_letter
import requests
import io
```

# 1. ФУНКЦИЯ ДЛЯ ЗАГРУЗКИ С ЯНДЕКС ДИСКА


```python
def download_from_yadisk(public_url):
    """
    Загружает файл с Яндекс Диска по публичной ссылке
    """
    try:
        # Преобразуем публичную ссылку в ссылку для скачивания
        if 'disk.yandex.ru/d/' in public_url:
            file_id = public_url.split('disk.yandex.ru/d/')[-1].split('/')[0].split('?')[0]
            download_url = f"https://cloud-api.yandex.net/v1/disk/public/resources/download?public_key=https://disk.yandex.ru/d/{file_id}"
        else:
            download_url = f"https://cloud-api.yandex.net/v1/disk/public/resources/download?public_key={public_url}"

        # Получаем прямую ссылку для скачивания
        response = requests.get(download_url)
        response.raise_for_status()

        download_info = response.json()
        direct_download_url = download_info['href']

        # Скачиваем файл
        file_response = requests.get(direct_download_url)
        file_response.raise_for_status()

        return io.BytesIO(file_response.content)
    except Exception as e:
        print(f"Ошибка при загрузке файла: {e}")
        return None
```

# 2. ЗАГРУЗКА ДАННЫХ С ЯНДЕКС ДИСКА


```python
# Публичные ссылки на файлы
financial_url = "https://disk.yandex.ru/d/P3XoPAkqZnKpCw"
prolongations_url = "https://disk.yandex.ru/d/9r5LF3YyCr7p6g"

financial_data = download_from_yadisk(financial_url)
if financial_data:
    financial = pd.read_csv(financial_data)
else:
    try:
        financial_direct = "https://disk.yandex.ru/d/P3XoPAkqZnKpCw/download"
        response = requests.get(financial_direct)
        financial = pd.read_csv(io.BytesIO(response.content))
    except Exception as e:
        raise Exception(f"Не удалось загрузить financial_data.csv: {e}")

prolongations_data = download_from_yadisk(prolongations_url)
if prolongations_data:
    prolongations = pd.read_csv(prolongations_data)
else:
    try:
        prolongations_direct = "https://disk.yandex.ru/d/9r5LF3YyCr7p6g/download"
        response = requests.get(prolongations_direct)
        prolongations = pd.read_csv(io.BytesIO(response.content))
    except Exception as e:
        raise Exception(f"Не удалось загрузить prolongations.csv: {e}")
```

# 3. ПОДГОТОВКА ДАННЫХ

```python
financial.columns = [col.strip() for col in financial.columns]
prolongations.columns = [col.strip() for col in prolongations.columns]


def normalize_month_name(m):
    """
    Нормализация названий месяцев для единообразия
    """
    parts = str(m).strip().split()
    if len(parts) == 2:
        month, year = parts
        month = month.capitalize()
        month_mapping = {
            'Янв': 'Январь', 'Фев': 'Февраль', 'Мар': 'Март', 'Апр': 'Апрель',
            'Май': 'Май', 'Июн': 'Июнь', 'Июл': 'Июль', 'Август': 'Август',
            'Сен': 'Сентябрь', 'Окт': 'Октябрь', 'Ноя': 'Ноябрь', 'Дек': 'Декабрь'
        }
        for short, full in month_mapping.items():
            if month.startswith(short):
                month = full
                break
        return f"{month} {year}"
    return m


# Нормализация месяца в prolongations
prolongations['month'] = prolongations['month'].apply(normalize_month_name)
```

# 4. ОБЪЕДИНЕНИЕ ДАННЫХ


```python
df = financial.merge(prolongations[['id', 'month', 'AM']], on='id', how='left')
```

# 5. ФИЛЬТРАЦИЯ ПО 2023 ГОДУ И УДАЛЕНИЕ БЕЗ А/М


```python
# Находим все месячные колонки за 2023 год
month_cols_2023 = [
    col for col in df.columns
    if '2023' in col and not any(x in col.lower() for x in ['id', 'причина', 'account', 'month', 'am'])
]

# Порядок месяцев для сортировки
months_order = {
    'январь': 1, 'февраль': 2, 'март': 3, 'апрель': 4,
    'май': 5, 'июнь': 6, 'июль': 7, 'август': 8,
    'сентябрь': 9, 'октябрь': 10, 'ноябрь': 11, 'декабрь': 12
}


def parse_month(label):
    """
    Функция для парсинга и сортировки месяцев
    """
    parts = str(label).split()
    if len(parts) == 2:
        month, year = parts
        return int(year), months_order.get(month.lower(), 0)
    return (9999, 99)


# Сортируем месяцы в правильном порядке
month_cols_2023 = sorted(month_cols_2023, key=parse_month)
```

# 6. ФУНКЦИЯ ОЧИСТКИ СУММ

```python
def clean_amount(s):
    """
    Функция для очистки и преобразования денежных сумм
    """
    if pd.isna(s):
        return 0.0
    s = str(s)
    s = re.sub(r'\s+|\xa0', '', s)
    s = s.replace(',', '.')
    if s.lower().strip() in ['стоп', 'stop', '']:
        return 0.0
    try:
        return float(s)
    except:
        return 0.0
```

# 7. РАСЧЕТ КОЭФФИЦИЕНТОВ


```python
# Собираем текстовый отчет
text_report = []

# Очищаем данные
df_clean = df.copy()
for col in month_cols_2023:
    df_clean[col] = df_clean[col].apply(clean_amount)

# ФИЛЬТРУЕМ ТОЛЬКО С АККАУНТ-МЕНЕДЖЕРАМИ
mask_am = (df_clean['AM'].notna()) & \
          (df_clean['AM'] != '') & \
          (df_clean['AM'].str.strip().str.lower() != 'без а/м')
df_clean = df_clean[mask_am]

# Получаем список менеджеров с А/М
managers_with_am = [
    am for am in df_clean['AM'].dropna().unique()
    if am and str(am).strip() != '' and str(am).strip().lower() != 'без а/м'
]

# === РАСЧЕТЫ ПО МЕСЯЦАМ ===
# Словари для хранения результатов
monthly_results_managers = {}  # По менеджерам по месяцам
monthly_results_department = {}  # По отделу по месяцам

for i in range(2, len(month_cols_2023)):
    current_month = month_cols_2023[i]  # Текущий месяц (например, "Март 2023")
    prev_month = month_cols_2023[i - 1]  # Предыдущий месяц (например, "Февраль 2023")
    prev_prev_month = month_cols_2023[i - 2]  # Позапрошлый месяц (например, "Январь 2023")

    short_current = current_month.split()[0]  # Короткое название текущего месяца

    # Расчет по отделу в целом
    dept_data = df_clean

    # Коэффициент 1 для отдела
    dept_finished_prev = dept_data[dept_data['month'] == prev_month]
    dept_sum_finished_prev = dept_finished_prev[prev_month].sum()
    dept_sum_prolong_first = dept_finished_prev[current_month].sum()
    dept_coef_first = dept_sum_prolong_first / dept_sum_finished_prev if dept_sum_finished_prev > 0 else 0

    # Коэффициент 2 для отдела
    dept_finished_prev2 = dept_data[dept_data['month'] == prev_prev_month]
    dept_no_prolong_first = dept_finished_prev2[dept_finished_prev2[prev_month] == 0]
    dept_sum_finished_prev2 = dept_no_prolong_first[prev_prev_month].sum()
    dept_sum_prolong_second = dept_no_prolong_first[current_month].sum()
    dept_coef_second = dept_sum_prolong_second / dept_sum_finished_prev2 if dept_sum_finished_prev2 > 0 else 0

    # Сохраняем результаты по отделу
    monthly_results_department[short_current] = {
        'Коэф. 1-го месяца': dept_coef_first,
        'Коэф. 2-го месяца': dept_coef_second,
        'Сумма завершенных (1-й месяц)': dept_sum_finished_prev,
        'Сумма пролонгированных (1-й месяц)': dept_sum_prolong_first,
        'Сумма завершенных (2-й месяц)': dept_sum_finished_prev2,
        'Сумма пролонгированных (2-й месяц)': dept_sum_prolong_second
    }

    # Расчет по каждому менеджеру
    monthly_results_managers[short_current] = {}

    for am in managers_with_am:
        am_data = df_clean[df_clean['AM'] == am]

        # Коэффициент 1 для менеджера
        projects_finished_prev = am_data[am_data['month'] == prev_month]
        sum_finished_prev = projects_finished_prev[prev_month].sum()
        sum_prolong_first = projects_finished_prev[current_month].sum()
        coef_first = sum_prolong_first / sum_finished_prev if sum_finished_prev > 0 else 0

        # Коэффициент 2 для менеджера
        projects_finished_prev2 = am_data[am_data['month'] == prev_prev_month]
        no_prolong_first = projects_finished_prev2[projects_finished_prev2[prev_month] == 0]
        sum_finished_prev2 = no_prolong_first[prev_prev_month].sum()
        sum_prolong_second = no_prolong_first[current_month].sum()
        coef_second = sum_prolong_second / sum_finished_prev2 if sum_finished_prev2 > 0 else 0

        # Сохраняем результаты по менеджеру
        monthly_results_managers[short_current][am] = {
            'Коэф. 1-го месяца': coef_first,
            'Коэф. 2-го месяца': coef_second,
            'Сумма завершенных (1-й месяц)': sum_finished_prev,
            'Сумма пролонгированных (1-й месяц)': sum_prolong_first,
            'Сумма завершенных (2-й месяц)': sum_finished_prev2,
            'Сумма пролонгированных (2-й месяц)': sum_prolong_second
        }

# === РАСЧЕТЫ ЗА ГОД ===
# Результаты за год
yearly_results_managers = {}  # По менеджерам за год
yearly_results_department = {}  # По отделу за год

# Расчет за год для отдела
dept_total_finished_first = sum([data['Сумма завершенных (1-й месяц)'] for data in monthly_results_department.values()])
dept_total_prolong_first = sum(
    [data['Сумма пролонгированных (1-й месяц)'] for data in monthly_results_department.values()])
dept_yearly_coef_first = dept_total_prolong_first / dept_total_finished_first if dept_total_finished_first > 0 else 0

dept_total_finished_second = sum(
    [data['Сумма завершенных (2-й месяц)'] for data in monthly_results_department.values()])
dept_total_prolong_second = sum(
    [data['Сумма пролонгированных (2-й месяц)'] for data in monthly_results_department.values()])
dept_yearly_coef_second = dept_total_prolong_second / dept_total_finished_second if dept_total_finished_second > 0 else 0

dept_total_finished = dept_total_finished_first + dept_total_finished_second
dept_total_prolong = dept_total_prolong_first + dept_total_prolong_second
dept_total_coef = dept_total_prolong / dept_total_finished if dept_total_finished > 0 else 0

yearly_results_department = {
    'Коэф. пролонгации 1-го месяца': dept_yearly_coef_first,
    'Коэф. пролонгации 2-го месяца': dept_yearly_coef_second,
    'Общий коэффициент пролонгации': dept_total_coef,
    'Сумма завершенных (1-й месяц)': dept_total_finished_first,
    'Сумма пролонгированных (1-й месяц)': dept_total_prolong_first,
    'Сумма завершенных (2-й месяц)': dept_total_finished_second,
    'Сумма пролонгированных (2-й месяц)': dept_total_prolong_second,
    'Общая сумма завершенных': dept_total_finished,
    'Общая сумма пролонгированных': dept_total_prolong
}

# Расчет за год для каждого менеджера
for am in managers_with_am:
    total_finished_first = 0
    total_prolong_first = 0
    total_finished_second = 0
    total_prolong_second = 0

    for month_data in monthly_results_managers.values():
        if am in month_data:
            manager_data = month_data[am]
            total_finished_first += manager_data['Сумма завершенных (1-й месяц)']
            total_prolong_first += manager_data['Сумма пролонгированных (1-й месяц)']
            total_finished_second += manager_data['Сумма завершенных (2-й месяц)']
            total_prolong_second += manager_data['Сумма пролонгированных (2-й месяц)']

    yearly_coef_first = total_prolong_first / total_finished_first if total_finished_first > 0 else 0
    yearly_coef_second = total_prolong_second / total_finished_second if total_finished_second > 0 else 0

    total_finished = total_finished_first + total_finished_second
    total_prolong = total_prolong_first + total_prolong_second
    total_coef = total_prolong / total_finished if total_finished > 0 else 0

    yearly_results_managers[am] = {
        'Коэф. пролонгации 1-го месяца': yearly_coef_first,
        'Коэф. пролонгации 2-го месяца': yearly_coef_second,
        'Общий коэффициент пролонгации': total_coef,
        'Сумма завершенных (1-й месяц)': total_finished_first,
        'Сумма пролонгированных (1-й месяц)': total_prolong_first,
        'Сумма завершенных (2-й месяц)': total_finished_second,
        'Сумма пролонгированных (2-й месяц)': total_prolong_second,
        'Общая сумма завершенных': total_finished,
        'Общая сумма пролонгированных': total_prolong
    }
```

# 8. СОЗДАНИЕ ФОРМАТИРОВАННОГО EXCEL ОТЧЕТА


```python
text_report.append("АНАЛИТИЧЕСКИЙ ОТЧЕТ ПО ПРОЛОНГАЦИЯМ ЗА 2023 ГОД")
text_report.append("")
text_report.append("ОПИСАНИЕ РАСЧЕТОВ:")
text_report.append("Коэффициент 1-го месяца: отношение суммы отгрузки проектов, пролонгированных")
text_report.append("в первый месяц после завершения к сумме отгрузки завершившихся в прошлом месяце проектов")
text_report.append("Коэффициент 2-го месяца: отношение суммы отгрузки проектов, пролонгированных")
text_report.append("во второй месяц к сумме отгрузки проектов, не пролонгированных в первый месяц")
text_report.append("")
text_report.append("РАСЧЕТЫ ВКЛЮЧАЮТ:")
text_report.append("1. Помесячные коэффициенты для каждого менеджера")
text_report.append("2. Помесячные коэффициенты для отдела в целом")
text_report.append("3. Годовые коэффициенты для каждого менеджера")
text_report.append("4. Годовые коэффициенты для отдела в целом")
text_report.append("")

# Добавляем статистику
text_report.append("КЛЮЧЕВАЯ СТАТИСТИКА ЗА 2023 ГОД:")
text_report.append(
    f"Общий коэффициент пролонгации отдела: {yearly_results_department['Общий коэффициент пролонгации']:.2%}")
text_report.append(f"Коэффициент 1-го месяца отдела: {yearly_results_department['Коэф. пролонгации 1-го месяца']:.2%}")
text_report.append(f"Коэффициент 2-го месяца отдела: {yearly_results_department['Коэф. пролонгации 2-го месяца']:.2%}")
text_report.append(
    f"Общая сумма завершенных проектов: {yearly_results_department['Общая сумма завершенных']:,.0f} руб.")
text_report.append(
    f"Общая сумма пролонгированных: {yearly_results_department['Общая сумма пролонгированных']:,.0f} руб.")
text_report.append(f"Количество менеджеров: {len(managers_with_am)}")

# Находим лучшего менеджера
if yearly_results_managers:
    best_manager = max(yearly_results_managers.items(), key=lambda x: x[1]['Общий коэффициент пролонгации'])
    text_report.append(
        f"Лучший менеджер: {best_manager[0]} (коэффициент {best_manager[1]['Общий коэффициент пролонгации']:.2%})")

# === 9. СОЗДАНИЕ ФОРМАТИРОВАННОГО EXCEL ОТЧЕТА ===
wb = Workbook()

# Стили
header_fill = PatternFill(start_color="366092", end_color="366092", fill_type="solid")
header_font = Font(color="FFFFFF", bold=True)
border = Border(left=Side(style='thin'), right=Side(style='thin'),
                top=Side(style='thin'), bottom=Side(style='thin'))

# === ЛИСТ 1: ТЕКСТОВЫЙ ОТЧЕТ ===
ws1 = wb.active
ws1.title = "Текстовый отчет"

ws1.merge_cells('A1:D1')
ws1['A1'] = "АНАЛИТИЧЕСКИЙ ОТЧЕТ ПО ПРОЛОНГАЦИЯМ ЗА 2023 ГОД"
ws1['A1'].font = Font(size=16, bold=True)
ws1['A1'].alignment = Alignment(horizontal='center')

# Записываем текстовый отчет
for row, line in enumerate(text_report, 3):
    ws1.cell(row=row, column=1, value=line)
    if line.startswith("КЛЮЧЕВАЯ СТАТИСТИКА"):
        ws1.cell(row=row, column=1).font = Font(bold=True, size=12)
    elif line.startswith("Общий коэффициент") or line.startswith("Коэффициент") or line.startswith(
            "Общая сумма") or line.startswith("Количество") or line.startswith("Лучший менеджер"):
        ws1.cell(row=row, column=1).font = Font(size=11)
    elif line.startswith("1.") or line.startswith("2.") or line.startswith("3.") or line.startswith("4."):
        ws1.cell(row=row, column=1).font = Font(bold=True)

# === ЛИСТ 2: СВОДНЫЙ ОТЧЕТ ЗА ГОД ===
ws2 = wb.create_sheet("Сводный отчет за год")

# Заголовок
ws2.merge_cells('A1:J1')
title_cell = ws2['A1']
title_cell.value = "СВОДНЫЙ ОТЧЕТ ПО МЕНЕДЖЕРАМ ЗА 2023 ГОД"
title_cell.font = Font(size=16, bold=True)
title_cell.alignment = Alignment(horizontal='center')

# Заголовки таблицы
headers = ['Менеджер', 'Общий коэф. пролонгации', 'Коэф. 1-го месяца', 'Коэф. 2-го месяца',
           'Сумма завершенных', 'Сумма пролонгированных', 'Эффективность']
for col, header in enumerate(headers, 1):
    cell = ws2.cell(row=3, column=col, value=header)
    cell.fill = header_fill
    cell.font = header_font
    cell.border = border
    cell.alignment = Alignment(horizontal='center')

# Данные менеджеров за год
row = 4
for am, data in sorted(yearly_results_managers.items(), key=lambda x: x[1]['Общий коэффициент пролонгации'],
                       reverse=True):
    ws2.cell(row=row, column=1, value=am).border = border
    ws2.cell(row=row, column=2, value=data['Общий коэффициент пролонгации']).number_format = '0.00%'
    ws2.cell(row=row, column=3, value=data['Коэф. пролонгации 1-го месяца']).number_format = '0.00%'
    ws2.cell(row=row, column=4, value=data['Коэф. пролонгации 2-го месяца']).number_format = '0.00%'
    ws2.cell(row=row, column=5, value=data['Общая сумма завершенных']).number_format = '#,##0'
    ws2.cell(row=row, column=6, value=data['Общая сумма пролонгированных']).number_format = '#,##0'

    efficiency = data['Общий коэффициент пролонгации']
    efficiency_cell = ws2.cell(row=row, column=7, value=efficiency)
    efficiency_cell.number_format = '0.00%'

    # Цветовое кодирование эффективности
    if efficiency >= 0.8:
        efficiency_cell.fill = PatternFill(start_color="C6EFCE", end_color="C6EFCE", fill_type="solid")
    elif efficiency >= 0.5:
        efficiency_cell.fill = PatternFill(start_color="FFEB9C", end_color="FFEB9C", fill_type="solid")
    else:
        efficiency_cell.fill = PatternFill(start_color="FFC7CE", end_color="FFC7CE", fill_type="solid")

    for col in range(1, 8):
        ws2.cell(row=row, column=col).border = border
    row += 1

# Данные по отделу за год
ws2.cell(row=row, column=1, value='ВСЕГО ПО ОТДЕЛУ').border = border
ws2.cell(row=row, column=2, value=yearly_results_department['Общий коэффициент пролонгации']).number_format = '0.00%'
ws2.cell(row=row, column=3, value=yearly_results_department['Коэф. пролонгации 1-го месяца']).number_format = '0.00%'
ws2.cell(row=row, column=4, value=yearly_results_department['Коэф. пролонгации 2-го месяца']).number_format = '0.00%'
ws2.cell(row=row, column=5, value=yearly_results_department['Общая сумма завершенных']).number_format = '#,##0'
ws2.cell(row=row, column=6, value=yearly_results_department['Общая сумма пролонгированных']).number_format = '#,##0'

efficiency = yearly_results_department['Общий коэффициент пролонгации']
efficiency_cell = ws2.cell(row=row, column=7, value=efficiency)
efficiency_cell.number_format = '0.00%'

for col in range(1, 8):
    ws2.cell(row=row, column=col).border = border

# График рейтинга менеджеров
if yearly_results_managers:
    chart1 = BarChart()
    chart1.type = "col"
    chart1.style = 10
    chart1.title = "Рейтинг менеджеров по эффективности пролонгации за год"
    chart1.y_axis.title = 'Коэффициент пролонгации'
    chart1.x_axis.title = 'Менеджеры'
    chart1.legend = None

    chart1.width = 18
    chart1.height = 12

    data = Reference(ws2, min_col=2, min_row=3, max_row=row - 1, max_col=2)
    categories = Reference(ws2, min_col=1, min_row=4, max_row=row - 1)
    chart1.add_data(data, titles_from_data=True)
    chart1.set_categories(categories)

    ws2.add_chart(chart1, "I3")

# === ЛИСТ 3: ДИНАМИКА ПО МЕСЯЦАМ (ОТДЕЛ) ===
ws3 = wb.create_sheet("Динамика по месяцам")

ws3.merge_cells('A1:E1')
ws3['A1'] = "ДИНАМИКА КОЭФФИЦИЕНТОВ ПРОЛОНГАЦИИ ПО МЕСЯЦАМ (ОТДЕЛ)"
ws3['A1'].font = Font(size=14, bold=True)
ws3['A1'].alignment = Alignment(horizontal='center')

headers = ['Месяц', 'Коэф. 1-го месяца', 'Коэф. 2-го месяца', 'Сумма завершенных', 'Сумма пролонгированных']
for col, header in enumerate(headers, 1):
    cell = ws3.cell(row=3, column=col, value=header)
    cell.fill = header_fill
    cell.font = header_font
    cell.border = border
    cell.alignment = Alignment(horizontal='center')

row = 4
for month, data in monthly_results_department.items():
    ws3.cell(row=row, column=1, value=month).border = border
    ws3.cell(row=row, column=2, value=data['Коэф. 1-го месяца']).number_format = '0.00%'
    ws3.cell(row=row, column=3, value=data['Коэф. 2-го месяца']).number_format = '0.00%'
    ws3.cell(row=row, column=4, value=data['Сумма завершенных (1-й месяц)']).number_format = '#,##0'
    ws3.cell(row=row, column=5, value=data['Сумма пролонгированных (1-й месяц)']).number_format = '#,##0'

    for col in range(1, 6):
        ws3.cell(row=row, column=col).border = border
    row += 1

# График динамики отдела
if monthly_results_department:
    chart2 = LineChart()
    chart2.title = "Динамика коэффициентов пролонгации отдела по месяцам"
    chart2.y_axis.title = 'Коэффициент'
    chart2.x_axis.title = 'Месяцы'

    chart2.width = 22
    chart2.height = 12

    data = Reference(ws3, min_col=2, min_row=3, max_row=row - 1, max_col=3)
    categories = Reference(ws3, min_col=1, min_row=4, max_row=row - 1)
    chart2.add_data(data, titles_from_data=True)
    chart2.set_categories(categories)

    chart2.legend.position = 'r'
    ws3.add_chart(chart2, "G2")

# === ЛИСТ 4: ДЕТАЛЬНЫЕ ДАННЫЕ ПО МЕНЕДЖЕРАМ ПО МЕСЯЦАМ ===
ws4 = wb.create_sheet("Детальные данные")

ws4.merge_cells('A1:H1')
ws4['A1'] = "ДЕТАЛЬНЫЕ ДАННЫЕ ПО ПРОЛОНГАЦИЯМ ПО МЕСЯЦАМ"
ws4['A1'].font = Font(size=14, bold=True)
ws4['A1'].alignment = Alignment(horizontal='center')

detailed_headers = ['Менеджер', 'Месяц', 'Коэф. 1-го месяца', 'Коэф. 2-го месяца',
                    'Сумма завершенных (1-й)', 'Сумма пролонгированных (1-й)',
                    'Сумма завершенных (2-й)', 'Сумма пролонгированных (2-й)']

for col, header in enumerate(detailed_headers, 1):
    cell = ws4.cell(row=3, column=col, value=header)
    cell.fill = header_fill
    cell.font = header_font
    cell.border = border
    cell.alignment = Alignment(horizontal='center')

row = 4
for month, managers_data in monthly_results_managers.items():
    for am, data in managers_data.items():
        ws4.cell(row=row, column=1, value=am).border = border
        ws4.cell(row=row, column=2, value=month).border = border
        ws4.cell(row=row, column=3, value=data['Коэф. 1-го месяца']).number_format = '0.00%'
        ws4.cell(row=row, column=4, value=data['Коэф. 2-го месяца']).number_format = '0.00%'
        ws4.cell(row=row, column=5, value=data['Сумма завершенных (1-й месяц)']).number_format = '#,##0'
        ws4.cell(row=row, column=6, value=data['Сумма пролонгированных (1-й месяц)']).number_format = '#,##0'
        ws4.cell(row=row, column=7, value=data['Сумма завершенных (2-й месяц)']).number_format = '#,##0'
        ws4.cell(row=row, column=8, value=data['Сумма пролонгированных (2-й месяц)']).number_format = '#,##0'

        for col in range(1, 9):
            ws4.cell(row=row, column=col).border = border
        row += 1

# === ЛИСТ 5: АНАЛИЗ ЭФФЕКТИВНОСТИ ===
ws5 = wb.create_sheet("Анализ эффективности")

ws5.merge_cells('A1:B1')
ws5['A1'] = "КЛЮЧЕВЫЕ МЕТРИКИ ОТДЕЛА ЗА 2023 ГОД"
ws5['A1'].font = Font(size=14, bold=True)
ws5['A1'].alignment = Alignment(horizontal='center')

metrics = [
    ("Общий коэффициент пролонгации 1-го месяца", yearly_results_department['Коэф. пролонгации 1-го месяца']),
    ("Общий коэффициент пролонгации 2-го месяца", yearly_results_department['Коэф. пролонгации 2-го месяца']),
    ("Общий коэффициент пролонгации", yearly_results_department['Общий коэффициент пролонгации']),
    ("Общая сумма завершенных проектов", yearly_results_department['Общая сумма завершенных']),
    ("Общая сумма пролонгированных", yearly_results_department['Общая сумма пролонгированных']),
    ("Эффективность пролонгации",
     yearly_results_department['Общая сумма пролонгированных'] / yearly_results_department['Общая сумма завершенных']),
    ("Количество менеджеров", len(managers_with_am)),
]

if yearly_results_managers:
    best_manager = max(yearly_results_managers.items(), key=lambda x: x[1]['Общий коэффициент пролонгации'])
    metrics.extend([
        ("Лучший менеджер", best_manager[0]),
        ("Эффективность лучшего менеджера", best_manager[1]['Общий коэффициент пролонгации']),
        ("Сумма пролонгаций лучшего менеджера", best_manager[1]['Общая сумма пролонгированных'])
    ])

for row, (metric, value) in enumerate(metrics, 3):
    ws5.cell(row=row, column=1, value=metric).font = Font(bold=True)
    ws5.cell(row=row, column=1).border = border

    value_cell = ws5.cell(row=row, column=2, value=value)
    value_cell.border = border

    if isinstance(value, (int, float)) and value < 1:
        value_cell.number_format = '0.00%'
        if value >= 0.7:
            value_cell.fill = PatternFill(start_color="C6EFCE", end_color="C6EFCE", fill_type="solid")
        elif value >= 0.5:
            value_cell.fill = PatternFill(start_color="FFEB9C", end_color="FFEB9C", fill_type="solid")
        else:
            value_cell.fill = PatternFill(start_color="FFC7CE", end_color="FFC7CE", fill_type="solid")
    elif isinstance(value, (int, float)):
        value_cell.number_format = '#,##0'

# Круговой график эффективности
if yearly_results_managers:
    chart3 = PieChart()
    chart3.title = "Распределение пролонгаций по менеджерам за год"

    chart3.width = 16
    chart3.height = 10

    labels = Reference(ws2, min_col=1, min_row=4, max_row=4 + len(yearly_results_managers) - 1)
    data = Reference(ws2, min_col=6, min_row=4, max_row=4 + len(yearly_results_managers) - 1)
    chart3.add_data(data, titles_from_data=True)
    chart3.set_categories(labels)

    chart3.legend.position = 'r'
    ws5.add_chart(chart3, "D3")


# === НАСТРОЙКА ШИРИНЫ КОЛОНОК ===
def auto_adjust_column_width(ws):
    for col in range(1, ws.max_column + 1):
        max_length = 0
        col_letter = get_column_letter(col)
        for row in range(1, ws.max_row + 1):
            cell = ws.cell(row=row, column=col)
            if cell.value:
                try:
                    if len(str(cell.value)) > max_length:
                        max_length = len(str(cell.value))
                except:
                    pass
        adjusted_width = min(50, (max_length + 2))
        ws.column_dimensions[col_letter].width = adjusted_width


# Применяем настройку ширины ко всем листам
for ws in [ws1, ws2, ws3, ws4, ws5]:
    auto_adjust_column_width(ws)

# Сохраняем файл
wb.save("Аналитика_пролонгаций_отчет.xlsx")
```
